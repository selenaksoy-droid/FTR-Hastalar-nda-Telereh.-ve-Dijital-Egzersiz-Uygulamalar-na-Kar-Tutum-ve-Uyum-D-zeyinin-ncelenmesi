```python
import pandas as pd
import numpy as np

# Örnek Veri Seti Oluşturma (Telerehabilitasyon Hastaları)
np.random.seed(42)
hasta_sayisi = 50

veri = {
    'Hasta_ID': range(1, hasta_sayisi + 1),
    'Yas': np.random.randint(20, 75, size=hasta_sayisi),
    'Egzersiz_Uyum_Orani_Yuzde': np.random.randint(40, 100, size=hasta_sayisi),
    'Teknoloji_Kabul_Puani_1_5': np.random.uniform(2.0, 5.0, size=hasta_sayisi).round(1)
}

df = pd.DataFrame(veri)

print("--- TELEREHABİLİTASYON HASTA UYUM VE TUTUM ANALİZİ ---")
print(df.head(10)) # İlk 10 hastayı göster
print("\n--- GENEL İSTATİSTİKLER ---")
print(f"Ortalama Egzersiz Uyum Oranı: %{df['Egzersiz_Uyum_Orani_Yuzde'].mean():.2f}")
print(f"Ortalama Teknoloji Kabul (Tutum) Puanı (1-5): {df['Teknoloji_Kabul_Puani_1_5'].mean():.2f}")
```
